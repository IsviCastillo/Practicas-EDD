## 2. MÉTODOS CLAVE

```python
inventario = {
    "TECLADO": 15,
    "MOUSE": 28,
    "MONITOR": 8,
    "HEADSET": 12
}
```

```python
print("\n--- Catálogo de Productos ---")
for codigo, stock in inventario.items():
    print(f"Producto: {codigo:<10} | Stock: {stock}")
```

Operaciones de actualización y eliminación

```python
# Modificación
inventario["MOUSE"] += 5

#Inserción
inventario["WEBCAM"] = 10

# Eliminación segura
eliminado = inventario.pop("MONITOR", None)

print(f"Eliminado: {eliminado} unidades de MONITOR")
```

```python
print("\n--- Catálogo de Productos ---")
for codigo, stock in inventario.items():
    print(f"Producto: {codigo:<10} | Stock: {stock}")
```

## 3. CONTEO DE FRECUENCIAS (PATRÓN HASH MAP)

```python
accesos_servidor = ["192.168.1.1", "10.0.0.2", "192.168.1.1", "172.16.0.5", "192.168.1.1", "10.0.0.2"]
conteo_ips = {}

for ip in accesos_servidor:
    conteo_ips[ip] = conteo_ips.get(ip, 0) + 1

print("\nFrecuencia de accesos por IP:", conteo_ips)
```
