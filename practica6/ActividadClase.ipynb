## Reto: Sistema de Monitoreo e Inventario Hash

## 1. Mapeo de registros por SKU


```python
inventario = {
    "SKU-01": {
        "nombre": "Auriculares",
        "precio": 44.99,
        "existencia": 15
    },
    "SKU-02": {
        "nombre": "Nintendo Switch",
        "precio": 450.50,
        "existencia": 8
    },
    "SKU-03": {
        "nombre": "Mouse Xiaomi",
        "precio": 24.99,
        "existencia": 18
    }
}

print(":.Inventario por SKU.:")
for sku, producto in inventario.items():
    print(f"{sku}: {producto['nombre']} | Precio: ${producto['precio']:.2f} | Existencia: {producto['existencia']}")
```

```python
sku_consultado = "SKU-02"
producto = inventario.get(sku_consultado)

if producto is not None:
    print(f"Producto consultado: {producto['nombre']}")
    print(f"Existencia disponible: {producto['existencia']}")
else:
    print("SKU no encontrado")
```

## 2. Contador de ocurrencias

```python
transacciones = ["entrada", "salida", "entrada", "ajuste","salida", "entrada", "salida", "entrada"]
frecuencia_transacciones = {}

for transaccion in transacciones:
    frecuencia_transacciones[transaccion] = frecuencia_transacciones.get(transaccion, 0) + 1

print(":.Frecuencia de transacciones.:")
for transaccion, cantidad in frecuencia_transacciones.items():
    print(f"{transaccion}: {cantidad}")
```

## 3. Agrupamiento por categoría


```python
productos = [
    ("Teclado mecánico", "Periféricos"),
    ("Mouse inalámbrico", "Periféricos"),
    ("Monitor HP", "Pantallas"),
    ("Monitor curvo Samsung", "Pantallas"),
    ("Cable HDMI", "Accesorios")
]

productos_por_categoria = {}

for nombre, categoria in productos:
    if categoria not in productos_por_categoria:
        productos_por_categoria[categoria] = []
    productos_por_categoria[categoria].append(nombre)

print(":.Productos agrupados por categoría.:")
for categoria, elementos in productos_por_categoria.items():
    print(f"{categoria}: {', '.join(elementos)}")
```
