# Práctica 5: Simplificación y Operaciones con Lambda

**Reto:** Refactorización, validaciones y procesamiento en lote.

```python
# 1. Refactorización a Lambda (Descuento, Temperatura, Nombre)
calcular_descuento = lambda precio, porcentaje: precio * (porcentaje / 100)
convertir_temperatura = lambda celsius: (celsius * 9/5) + 32
formatear_nombre = lambda nombre, apellido: f"{apellido.upper()}, {nombre.capitalize()}"

print("Descuento:", calcular_descuento(200, 15))
print("Temperatura:", convertir_temperatura(25))
print("Nombre:", formatear_nombre("isvi", "castillo"))
```

```python
# 2. Validaciones en una línea (Positivo/Negativo, Cero y Mayoría de edad)
evaluar_signo = lambda n: "Positivo" if n > 0 else "Negativo" if n < 0 else "Cero"
es_mayor_edad = lambda edad: edad >= 18

print("Signo de 10:", evaluar_signo(10))
print("Signo de -5:", evaluar_signo(-5))
print("Signo de 0:", evaluar_signo(0))
print("¿Edad 20 es mayor?:", es_mayor_edad(20))
print("¿Edad 15 es mayor?:", es_mayor_edad(15))
```

```python
# 3. Procesamiento en lote (Aplicar recargo a lista de precios)
aplicar_recargo = lambda precio: round(precio * 1.10, 2)

precios = [100, 250, 50, 1200]
precios_con_recargo = [aplicar_recargo(p) for p in precios]

print("Precios originales:", [f"{precio:.2f}" for precio in precios])
print("Precios con recargo:", [f"{precio:.2f}" for precio in precios_con_recargo])
```
