Ejercicio 1: Burbuja Optimizado con Bandera de Intercambio
Optimice el algoritmo de Burbuja utilizando una variable booleana que detecte si en una pasada no hubo intercambios para finalizar tempranamente si la lista ya está ordenada.

```python
def burbuja_optimizado(lista):
    n = len(lista)
    for i in range(n):
        hubo_intercambio = False
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                hubo_intercambio = True
        if not hubo_intercambio:
            break
    return lista

print(burbuja_optimizado([5, 3, 8, 1, 9, 2]))
print(burbuja_optimizado([1, 2, 3, 4, 5]))
```

Ejercicio 2: Ordenamiento Descendente de Registros
Ajuste el algoritmo de Selección o Inserción para ordenar una lista de calificaciones en orden descendente (de mayor a menor).

```python
def seleccion_descendente(lista):
    n = len(lista)
    for i in range(n):
        max_idx = i
        for j in range(i + 1, n):
            if lista[j] > lista[max_idx]:
                max_idx = j
        lista[i], lista[max_idx] = lista[max_idx], lista[i]
    return lista

calificaciones = [7.5, 9.0, 6.0, 8.5, 10.0, 5.5]
print(seleccion_descendente(calificaciones))
```
