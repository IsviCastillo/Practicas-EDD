## Tema: Recursividad Práctica
Estructuras de Datos — Caso Base, Caso Recursivo y Pila de Ejecución



```python
# 1. Factorial y Suma Acumulada Recursiva

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

def suma_acumulada(n):
    if n == 1:
        return 1
    return n + suma_acumulada(n - 1)

print(factorial(5))
print(suma_acumulada(10))
```

```python
# 2. Recorrido Recursivo de Estructuras (Listas)

def sumar_lista(lista):
    if not lista:
        return 0
    return lista[0] + sumar_lista(lista[1:])

def invertir_cadena(cadena):
    if len(cadena) <= 1:
        return cadena
    return cadena[-1] + invertir_cadena(cadena[:-1])

print(sumar_lista([5, 10, 15, 20]))
print(invertir_cadena("Estructuras"))
```
