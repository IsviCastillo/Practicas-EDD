Ejercicio 1: Potencia y Conteo Regresivo
Implemente una función recursiva potencia(base, exponente) que calcule $base^{exponente}$ sin usar el operador ** ni bucles for/while.

```python
def potencia(base, exponente):
    if exponente == 0:
        return 1
    return base * potencia(base, exponente - 1)

print(potencia(2,5))
print(potencia(3,4))
```

Ejercicio 2: Contador de Dígitos de un Número
Cree una función recursiva contar_digitos(n) que retorne la cantidad de dígitos que integran un número entero positivo dividiendo entre 10 sucesivamente.

```python
def contar_digitos(n):
    if n < 10:
        return 1
    return 1 + contar_digitos(n // 10)

print(contar_digitos(5))
print(contar_digitos(4825))
print(contar_digitos(1000000))
```

Ejercicio 3: Búsqueda del Elemento Máximo en una Lista
Diseñe una función recursiva maximo_recursivo(lista) que compare la cabeza de la lista con el máximo obtenido de la sublista restante hasta determinar el valor mayor.

```python
def maximo_recursivo(lista):
    if len(lista) == 1:
        return lista[0]
    maximo_resto = maximo_recursivo(lista[1:])
    if lista[0] > maximo_resto:
        return lista[0]
    return maximo_resto

print(maximo_recursivo([3, 8, 2, 10, 5]))
print(maximo_recursivo([7]))
print(maximo_recursivo([-4, -1, -9]))
```
