# Práctica 5: Sistema Integral de Caja y Despacho
## Estructuras FIFO, LIFO y Sets

**Indicaciones:**
1. **FIFO (Fila en Caja):** Usar `collections.deque` con `popleft()`. Validar fila vacía.
2. **LIFO (Bolsa de Productos):** Usar pila con `pop()` (último en entrar, primero en salir).
3. **Sets (Cupones y Membresías):** Validar membresías vigentes y evitar cupones repetidos.

```python
from collections import deque

# FIFO: Clientes
fila_caja = deque()

# Sets: Membresias y cupones
membresias_vigentes = {"VIP", "Socio", "Plus"}
cupones_usados = set()

print("Estructuras inicializadas.")
```

```python
# Funciones

def agregar_cliente(nombre):
    fila_caja.append({"nombre": nombre, "bolsa": []})


def agregar_producto(nombre_cliente, producto, precio):
    for cliente in fila_caja:
        if cliente["nombre"] == nombre_cliente:
            cliente["bolsa"].append((producto, precio))
            return

    print(f"No se encontro al cliente {nombre_cliente} en la fila.")


def registrar_cupon(cupon):
    if cupon in cupones_usados:
        print(f"El cupon {cupon} ya fue usado.")
    else:
        cupones_usados.add(cupon)
        print(f"Cupon {cupon} registrado.")

#LIFO: Productos
def atender_cliente():
    if not fila_caja:
        print("No hay clientes en la fila.")
        return

    cliente = fila_caja.popleft()
    print(f"Atendiendo a: {cliente['nombre']}")

    total = 0
    while cliente["bolsa"]:
        producto, precio = cliente["bolsa"].pop()
        total += precio
        print(f"  {producto}: ${precio}")

    print(f"Total a pagar: ${total}")
```

```python
# Simulacion de operaciones
agregar_cliente("Ana")
agregar_cliente("Luis")

agregar_producto("Ana", "Pan", 1.50)
agregar_producto("Ana", "Leche", 2.00)
agregar_producto("Luis", "Huevos", 3.50)

registrar_cupon("PROMO2026")
registrar_cupon("PROMO2026")
registrar_cupon("PROMO2026")

print("Cupones usados:", cupones_usados)

print("Membresias VIP:", "VIP" in membresias_vigentes)
print("Membresias Socio:", "Socio" in membresias_vigentes)
print("Membresias Plus:", "Plus" in membresias_vigentes)

atender_cliente()
atender_cliente()
atender_cliente()
```
