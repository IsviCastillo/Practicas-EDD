```python
# Ejercicio 1: Árbol Genealógico Familiar

class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None
        self.der = None

raiz = Nodo("Abuelo")
raiz.izq = Nodo("Padre")
raiz.der = Nodo("Tío")

raiz.izq.izq = Nodo("Nieto 1")
raiz.izq.der = Nodo("Nieto 2")
raiz.der.izq = Nodo("Nieto 3")
raiz.der.der = Nodo("Nieto 4")

print("Raíz:", raiz.valor)
print("Hijos del abuelo:", raiz.izq.valor, "y", raiz.der.valor)
print("Nietos del padre:", raiz.izq.izq.valor, "y", raiz.izq.der.valor)
print("Nietos del tío:", raiz.der.izq.valor, "y", raiz.der.der.valor)


```

```python
# <!-- Ejercicio 2: Menú Interactivo de Decisiones

def preorden(nodo):
    if nodo is not None:
        print(nodo.valor, end=" ")
        preorden(nodo.izq)
        preorden(nodo.der)
def inorden(nodo):
    if nodo is not None:
        inorden(nodo.izq)
        print(nodo.valor, end=" ")
        inorden(nodo.der)
def postorden(nodo):
    if nodo is not None:
        postorden(nodo.izq)
        postorden(nodo.der)
        print(nodo.valor, end=" ")

decision = Nodo("¿El equipo enciende?")
 
decision.izq = Nodo("Sí: ¿Hay conexión a internet?")
decision.der = Nodo("No: Conectar el cable de corriente")
 
decision.izq.izq = Nodo("Sí: Cerrar ticket, problema resuelto")
decision.izq.der = Nodo("No: Reiniciar el router")
 
print("Preorden (orden en que se evalúan las opciones, de la raíz a las hojas):")
preorden(decision)
print()
 
print("Postorden (cierre de los procesos, de las hojas a la raíz):")
postorden(decision)
print()
```

```python
# Ejercicio 3: Reporte de Auditoría de Carpetas


def contar_nodos(nodo):
    if nodo is None:
        return 0
    return 1 + contar_nodos(nodo.izq) + contar_nodos(nodo.der)

def contar_hojas(nodo):
    if nodo is None:
        return 0
    if nodo.izq is None and nodo.der is None:
        return 1
    return contar_hojas(nodo.izq) + contar_hojas(nodo.der)
def buscar_valor(nodo, objetivo):
    if nodo is None:
        return False
    if nodo.valor == objetivo:
        return True
    return buscar_valor(nodo.izq, objetivo) or buscar_valor(nodo.der, objetivo)
carpetas = Nodo("Documentos")
 
carpetas.izq = Nodo("Trabajo")
carpetas.der = Nodo("Personal")
 
carpetas.izq.izq = Nodo("Informe.docx")
carpetas.izq.der = Nodo("Presupuesto.xlsx")
 
carpetas.der.izq = Nodo("Fotos")
carpetas.der.der = Nodo("CV.pdf")
 
carpetas.der.izq.izq = Nodo("Vacaciones.jpg")
 
print("Total de nodos (carpetas y archivos):", contar_nodos(carpetas))
print("Total de hojas (archivos finales):", contar_hojas(carpetas))
 
# El nombre del archivo lo ingresa el usuario
archivo = input("Ingrese el nombre del archivo a buscar: ")
if buscar_valor(carpetas, archivo):
    print(f"El archivo '{archivo}' SÍ existe en el árbol.")
else:
    print(f"El archivo '{archivo}' NO existe en el árbol.")
```
