Ejercicio 1: Planificador de Turnos de Procesador
Implemente una ListaCircularSimple para simular un planificador del sistema operativo. Agregue 4 procesos y realice un recorrido que simule 3 vueltas completas de ejecución continua.

```python
class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

class ListaCircularSimple:
    def __init__(self):
        self.cabeza = None

    def insertar(self, dato):
        nuevo_nodo = Nodo(dato)
        if not self.cabeza:
            self.cabeza = nuevo_nodo
            nuevo_nodo.siguiente = self.cabeza
        else:
            actual = self.cabeza
            while actual.siguiente != self.cabeza:
                actual = actual.siguiente
            actual.siguiente = nuevo_nodo
            nuevo_nodo.siguiente = self.cabeza

    def ejecutar_vueltas(self, vueltas):
        if not self.cabeza:
            return
        actual = self.cabeza
        for v in range(1, vueltas + 1):
            print(f"Vuelta {v}:")
            while True:
                print(f"  Ejecutando {actual.dato}")
                actual = actual.siguiente
                if actual == self.cabeza:
                    break

planificador = ListaCircularSimple()
planificador.insertar("Proceso 1")
planificador.insertar("Proceso 2")
planificador.insertar("Proceso 3")
planificador.insertar("Proceso 4")

planificador.ejecutar_vueltas(3)
```

## Ejercicio 2: Carrusel Infinito de Imágenes
Utilice la ListaCircularDoble para simular la vista previa de un carrusel gráfico. Implemente métodos para avanzar y retroceder de manera ilimitada verificando que el puntero no colapse en bordes.

```python
class NodoDoble:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None
        self.anterior = None

class ListaCircularDoble:
    def __init__(self):
        self.cabeza = None
        self.actual = None

    def insertar(self, dato):
        nuevo = NodoDoble(dato)
        if not self.cabeza:
            self.cabeza = nuevo
            nuevo.siguiente = nuevo
            nuevo.anterior = nuevo
            self.actual = nuevo
        else:
            cola = self.cabeza.anterior
            cola.siguiente = nuevo
            nuevo.anterior = cola
            nuevo.siguiente = self.cabeza
            self.cabeza.anterior = nuevo

    def avanzar(self):
        if self.actual:
            self.actual = self.actual.siguiente
            print(f"Avanzar  -> {self.actual.dato}")

    def retroceder(self):
        if self.actual:
            self.actual = self.actual.anterior
            print(f"Retroceder -> {self.actual.dato}")

carrusel = ListaCircularDoble()
carrusel.insertar("Imagen 1")
carrusel.insertar("Imagen 2")
carrusel.insertar("Imagen 3")

print("Imagen actual:", carrusel.actual.dato)

for i in range(4):
    carrusel.avanzar()

for i in range(4):
    carrusel.retroceder()
```
