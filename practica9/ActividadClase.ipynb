```python
# Ejercicio 1: Historial de Navegación Simple

class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None


class ListaEnlazada:
    def __init__(self):
        self.cabeza = None

    def insertar_final(self, dato):
        nuevo_nodo = Nodo(dato)

        if self.cabeza is None:
            self.cabeza = nuevo_nodo
        else:
            actual = self.cabeza
            while actual.siguiente:
                actual = actual.siguiente
            actual.siguiente = nuevo_nodo

    def imprimir(self):
        actual = self.cabeza
        while actual:
            print(actual.dato)
            actual = actual.siguiente


lista = ListaEnlazada()
lista.insertar_final("https://www.google.com")
lista.insertar_final("https://www.wikipedia.org")
lista.insertar_final("https://www.github.com")
lista.insertar_final("https://www.youtube.com")
lista.imprimir()  # Imprime el historial de navegación
```

```python
# Ejercicio 2: Cola de Pendientes y Despacho de Tareas

class ColaPendientes:
    def __init__(self):
        self.cabeza = None

    def eliminar_inicio(self):
        if self.cabeza is None:
            return None
        dato = self.cabeza.dato
        self.cabeza = self.cabeza.siguiente
        return dato

cola = ColaPendientes()
cola.cabeza = Nodo("Tarea 1")
cola.cabeza.siguiente = Nodo("Tarea 2")
cola.cabeza.siguiente.siguiente = Nodo("Tarea 3")    

# Simulación del procesamiento de tareas
while cola.cabeza:
    tarea_atendida = cola.eliminar_inicio()
    print(f"Tarea atendida: {tarea_atendida}")
    # Imprimir estado actual de la lista
    actual = cola.cabeza
    estado_actual = []
    while actual:
        estado_actual.append(actual.dato)
        actual = actual.siguiente
    print(f"Estado actual de la lista: {estado_actual}")
```

```python
# Ejercicio 3: Filtro y Conteo de Nodos en Memoria

def contar_elementos(self):
    actual = self.cabeza
    contador = 0
    while actual:
        contador += 1
        actual = actual.siguiente
    return contador

def obtener_posicion(self, indice):
    if indice < 0:
        return "Índice fuera de rango"
    actual = self.cabeza
    for i in range(indice):
        if actual is None:
            return "Índice fuera de rango"
        actual = actual.siguiente
    if actual is None:
        return "Índice fuera de rango"
    return actual.dato

# Incorporar los métodos a la clase, como en los ejemplos de la práctica
ColaPendientes.contar_elementos = contar_elementos
ColaPendientes.obtener_posicion = obtener_posicion

cola = ColaPendientes()
cola.cabeza = Nodo("Elemento 1")
cola.cabeza.siguiente = Nodo("Elemento 2")
cola.cabeza.siguiente.siguiente = Nodo("Elemento 3")
cola.cabeza.siguiente.siguiente.siguiente = Nodo("Elemento 4")
cola.cabeza.siguiente.siguiente.siguiente.siguiente = Nodo("Elemento 5")

# Contar elementos
total_elementos = cola.contar_elementos()

print(f"Total de elementos en la lista: {total_elementos}")

# Obtener elemento en el índice 2
elemento_en_indice_2 = cola.obtener_posicion(2)

print(f"Elemento en el índice 2: {elemento_en_indice_2}")
```
