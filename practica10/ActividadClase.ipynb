```python
# Ejercicio 1: Historial de Navegación con Botones Atrás/Adelante

class ListaDoblementeEnlazada:
    class Nodo:
        def __init__(self, url):
            self.url = url
            self.siguiente = None
            self.anterior = None

    def __init__(self):
        self.cabeza = None
        self.cola = None

    def agregar_url(self, url):
        nuevo_nodo = self.Nodo(url)
        if not self.cabeza:
            self.cabeza = nuevo_nodo
            self.cola = nuevo_nodo
        else:
            self.cola.siguiente = nuevo_nodo
            nuevo_nodo.anterior = self.cola
            self.cola = nuevo_nodo

    def recorrer_adelante(self):
        actual = self.cabeza
        while actual:
            print(actual.url)
            actual = actual.siguiente

    def recorrer_atras(self):
        actual = self.cola
        while actual:
            print(actual.url)
            actual = actual.anterior

# Crear una instancia de la lista doblemente enlazada
historial = ListaDoblementeEnlazada()
# Agregar 5 URLs al historial
urls = [
    "https://www.google.com",
    "https://www.github.com",
    "https://www.youtube.com",
    "https://www.canvas.com",
    "https://www.python.org"
]
for url in urls:
    historial.agregar_url(url)
    print(f"Agregado: {url}")

print("Recorriendo el historial hacia adelante:")
historial.recorrer_adelante()

print("Recorriendo el historial hacia atrás:")
historial.recorrer_atras()
```

```python
# Ejercicio 2: Reproductor de Música (Pistas musicales Siguiente y Anterior)

class ListaReproduccion:
    def __init__(self):
        self.cabeza = None
        self.cola = None

    def insertar_inicio(self, nombre):
        nuevo_nodo = ListaDoblementeEnlazada.Nodo(nombre)

        if self.cabeza is None:
            self.cabeza = nuevo_nodo
            self.cola = nuevo_nodo
        else:
            nuevo_nodo.siguiente = self.cabeza
            self.cabeza.anterior = nuevo_nodo
            self.cabeza = nuevo_nodo

    def eliminar_pista(self, nombre):
        actual = self.cabeza

        while actual is not None:
            if actual.url == nombre:
                if actual.anterior is not None:
                    actual.anterior.siguiente = actual.siguiente
                else:
                    self.cabeza = actual.siguiente

                if actual.siguiente is not None:
                    actual.siguiente.anterior = actual.anterior
                else:
                    self.cola = actual.anterior

                return True

            actual = actual.siguiente

        return False

reproduccion = ListaReproduccion()
pistas = [
    "Canción 1",
    "Canción 2",
    "Canción 3",
    "Canción 4",
    "Canción 5"
]

for pista in pistas:
    reproduccion.insertar_inicio(pista)
    print(f"Agregado: {pista}")

print(f"Pista eliminada: {reproduccion.eliminar_pista('Canción 3')}")

print("Recorriendo hacia adelante:")
actual = reproduccion.cabeza
while actual is not None:
    print(actual.url)
    actual = actual.siguiente

print("Recorriendo hacia atrás:")
actual = reproduccion.cola
while actual is not None:
    print(actual.url)
    actual = actual.anterior
```

```python
# Ejercicio 3: Búsqueda Bidireccional Óptima

def buscar_desde_cabeza(self, objetivo):
    actual = self.cabeza
    iteraciones = 0

    while actual is not None:
        iteraciones += 1
        if actual.url == objetivo:
            return True, iteraciones
        actual = actual.siguiente

    return False, iteraciones


def buscar_desde_cola(self, objetivo):
    actual = self.cola
    iteraciones = 0

    while actual is not None:
        iteraciones += 1
        if actual.url == objetivo:
            return True, iteraciones
        actual = actual.anterior

    return False, iteraciones

ListaDoblementeEnlazada.buscar_desde_cabeza = buscar_desde_cabeza
ListaDoblementeEnlazada.buscar_desde_cola = buscar_desde_cola

objetivo = "https://www.python.org"
encontrado_cabeza, iteraciones_cabeza = historial.buscar_desde_cabeza(objetivo)
encontrado_cola, iteraciones_cola = historial.buscar_desde_cola(objetivo)

print(f"Búsqueda desde la cabeza: encontrado={encontrado_cabeza}, iteraciones={iteraciones_cabeza}")
print(f"Búsqueda desde la cola: encontrado={encontrado_cola}, iteraciones={iteraciones_cola}")

if iteraciones_cola < iteraciones_cabeza:
    print("La búsqueda desde la cola fue más eficiente para este objetivo.")
else:
    print("La búsqueda desde la cabeza fue más eficiente para este objetivo.")
```
