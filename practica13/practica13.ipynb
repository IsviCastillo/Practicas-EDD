## Tema: Algoritmos de Búsqueda
Búsqueda Secuencial O(n) vs. Búsqueda Binaria O(log n)

## Código de Ejemplo

```python
# 1. Búsqueda Secuencial (Lineal)

def busqueda_secuencial(lista, objetivo):
    for i in range(len(lista)):
        if lista[i] == objetivo:
            return i
    return -1

datos = [14, 5, 8, 2, 99, 17]
print(busqueda_secuencial(datos, 99))
```

```python
# 2. Búsqueda Binaria (Iterativa)

def busqueda_binaria(lista, objetivo):
    inicio = 0
    fin = len(lista) - 1
    
    while inicio <= fin:
        medio = (inicio + fin) // 2
        if lista[medio] == objetivo:
            return medio
        elif lista[medio] < objetivo:
            inicio = medio + 1
        else:
            fin = medio - 1
    return -1

ordenados = [2, 5, 8, 14, 17, 99]
print(busqueda_binaria(ordenados, 17))
```
