Ejercicio 1: Contador de Pasos y Comparaciones
Modifique ambas funciones de búsqueda para que retornen una tupla (posicion, iteraciones). Pruebe buscar un número cercano al final en una lista ordenada de 1,000 elementos e imprima las iteraciones de cada método.

```python
def busqueda_secuencial(lista, objetivo):
    iteraciones = 0
    for i in range(len(lista)):
        iteraciones += 1
        if lista[i] == objetivo:
            return (i, iteraciones)
    return (-1, iteraciones)

def busqueda_binaria(lista, objetivo):
    inicio = 0
    fin = len(lista) - 1
    iteraciones = 0

    while inicio <= fin:
        iteraciones += 1
        medio = (inicio + fin) // 2
        if lista[medio] == objetivo:
            return (medio, iteraciones)
        elif lista[medio] < objetivo:
            inicio = medio + 1
        else:
            fin = medio - 1
    return (-1, iteraciones)

datos = list(range(1, 1001))
objetivo = 990

pos_sec, iter_sec = busqueda_secuencial(datos, objetivo)
pos_bin, iter_bin = busqueda_binaria(datos, objetivo)

print(f"Secuencial: posición {pos_sec}, iteraciones {iter_sec}")
print(f"Binaria:    posición {pos_bin}, iteraciones {iter_bin}")
```

Ejercicio 2: Búsqueda Binaria Recursiva
Implemente la versión recursiva busqueda_binaria_recursiva(lista, objetivo, inicio, fin) calculando los punteros móviles en cada llamada.

```python
def busqueda_binaria_recursiva(lista, objetivo, inicio, fin):
    if inicio > fin:
        return -1
    medio = (inicio + fin) // 2
    if lista[medio] == objetivo:
        return medio
    elif lista[medio] < objetivo:
        return busqueda_binaria_recursiva(lista, objetivo, medio + 1, fin)
    else:
        return busqueda_binaria_recursiva(lista, objetivo, inicio, medio - 1)

ordenados = [2, 5, 8, 14, 17, 99]

print(busqueda_binaria_recursiva(ordenados, 17, 0, len(ordenados) - 1))
print(busqueda_binaria_recursiva(ordenados, 2, 0, len(ordenados) - 1))
print(busqueda_binaria_recursiva(ordenados, 50, 0, len(ordenados) - 1))
```
