Ejercicio 1: Comparativa de Tiempos de Ejecución
Utilice el módulo time para medir la diferencia de tiempo al ordenar una lista aleatoria de 10,000 elementos mediante Burbuja frente a QuickSort.



```python
import time
import random

def ordenamiento_burbuja(lista):
    n = len(lista)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

def quicksort(lista):
    if len(lista) <= 1:
        return lista
    pivote = lista[len(lista) // 2]
    menores = [x for x in lista if x < pivote]
    iguales = [x for x in lista if x == pivote]
    mayores = [x for x in lista if x > pivote]
    return quicksort(menores) + iguales + quicksort(mayores)

datos = [random.randint(1, 100000) for _ in range(10000)]

copia_burbuja = datos.copy()
inicio = time.time()
ordenamiento_burbuja(copia_burbuja)
print("Burbuja:", time.time() - inicio)

copia_quick = datos.copy()
inicio = time.time()
quicksort(copia_quick)
print("QuickSort:", time.time() - inicio)
```

Ejercicio 2: Selección del Pivote en QuickSort
Modifique el algoritmo QuickSort para tomar como pivote el primer elemento de la lista y evalúe su comportamiento al ingresar una lista previamente ordenada.

```python
def quicksort_primer_pivote(lista):
    if len(lista) <= 1:
        return lista
    pivote = lista[0]
    menores = [x for x in lista if x < pivote]
    iguales = [x for x in lista if x == pivote]
    mayores = [x for x in lista if x > pivote]
    return quicksort_primer_pivote(menores) + iguales + quicksort_primer_pivote(mayores)

ordenada = list(range(1, 901))
inicio = time.perf_counter()
resultado = quicksort_primer_pivote(ordenada)
tiempo = time.perf_counter() - inicio

print("Lista ordenada correctamente:", resultado == ordenada)
print("Primeros elementos:", resultado[:5])
print("Tiempo con el primer elemento como pivote:", tiempo)
print("En una lista ordenada, el pivote siempre es el menor elemento.")
print("Por ello, una partición queda vacía y la otra contiene casi todos los elementos,\n"
      "degradando la complejidad de QuickSort a O(n^2).")
```
